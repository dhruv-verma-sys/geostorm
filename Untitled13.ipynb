```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset directly from IBM cloud storage
URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv'
df = pd.read_csv(URL)

print("Data loaded successfully!")
df.head()
```

```python
# Task 1.1: Line chart showing automobile sales fluctuation year to year
df_line = df.groupby('Year')['Automobile_Sales'].mean().reset_index()

plt.figure(figsize=(10, 6))
plt.plot(df_line['Year'], df_line['Automobile_Sales'], marker='o', color='b', linestyle='-')
plt.title('Automobile Sales Fluctuation Year to Year')
plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.xticks(df_line['Year'], rotation=45)
plt.grid(True)
plt.show()
```

```python
# Task 1.2: Ad expenditure vs. sales during non-recession periods
df_non_rec = df[df['Recession'] == 0]
df_ad_sales = df_non_rec.groupby('Year')[['Advertising_Expenditure', 'Automobile_Sales']].mean().reset_index()

fig, ax1 = plt.subplots(figsize=(10, 6))

color = 'tab:red'
ax1.set_xlabel('Year')
ax1.set_ylabel('Advertising Expenditure', color=color)
ax1.plot(df_ad_sales['Year'], df_ad_sales['Advertising_Expenditure'], color=color, marker='o')
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Automobile Sales', color=color)
ax2.plot(df_ad_sales['Year'], df_ad_sales['Automobile_Sales'], color=color, marker='s')
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Advertising Expenditure vs Automobile Sales during Non-Recession Periods')
plt.grid(True)
plt.show()
```

```python
# Task 1.3: Bar chart comparing sales per vehicle type during recession vs non-recession
df_bar = df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
df_bar['Recession'] = df_bar['Recession'].replace({0: 'Non-Recession', 1: 'Recession'})

plt.figure(figsize=(10, 6))
sns.barplot(x='Vehicle_Type', y='Automobile_Sales', hue='Recession', data=df_bar)
plt.title('Average Automobile Sales per Vehicle Type: Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Sales')
plt.show()
```

```python
# Task 1.4: Subplots for GDP variations
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]

fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(16, 6))

# Subplot 1: Recession GDP
sns.lineplot(x='Year', y='GDP', data=rec_data, ax=ax0, color='red', marker='o')
ax0.set_title('GDP Variation During Recession Periods')
ax0.set_xlabel('Year')
ax0.set_ylabel('GDP')

# Subplot 2: Non-Recession GDP
sns.lineplot(x='Year', y='GDP', data=non_rec_data, ax=ax1, color='green', marker='o')
ax1.set_title('GDP Variation During Non-Recession Periods')
ax1.set_xlabel('Year')
ax1.set_ylabel('GDP')

plt.tight_layout()
plt.show()
```

```python
# Task 1.5: Bubble plot for seasonality impact on sales
non_rec_data = df[df['Recession'] == 0]

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=non_rec_data,
    x='Month',
    y='Automobile_Sales',
    size='Seasonality_Weight',
    hue='Seasonality_Weight',
    sizes=(20, 400),
    palette='viridis'
)
plt.title('Impact of Seasonality on Automobile Sales (Non-Recession)')
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# Task 1.6: Scatter plot for average vehicle price vs sales volume during recessions
rec_data = df[df['Recession'] == 1]

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Price', y='Automobile_Sales', data=rec_data, color='purple', s=80)
plt.title('Average Vehicle Price vs Sales Volume During Recessions')
plt.xlabel('Price')
plt.ylabel('Automobile Sales Volume')
plt.grid(True)
plt.show()
```

```python
# Task 1.7: Total Ad Expenditure during Recession vs Non-Recession
ad_totals = df.groupby('Recession')['Advertising_Expenditure'].sum()

plt.figure(figsize=(8, 8))
plt.pie(
    ad_totals,
    labels=['Non-Recession', 'Recession'],
    autopct='%1.1f%%',
    startangle=90,
    colors=['#66b3ff', '#ff9999'],
    explode=(0, 0.1)
)
plt.title('Proportion of Advertising Expenditure: Non-Recession vs Recession')
plt.show()
```

```python
# Task 1.8: Total Ad Expenditure per vehicle type during Recession
rec_data = df[df['Recession'] == 1]
ad_by_type = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()

plt.figure(figsize=(8, 8))
plt.pie(
    ad_by_type,
    labels=ad_by_type.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel')
)
plt.title('Share of Advertising Expenditure by Vehicle Type During Recession')
plt.show()
```

```python
# Task 1.9: Unemployment rate vs vehicle type sales during recession
rec_data = df[df['Recession'] == 1]

plt.figure(figsize=(12, 6))
sns.lineplot(
    data=rec_data,
    x='unemployment_rate',
    y='Automobile_Sales',
    hue='Vehicle_Type',
    style='Vehicle_Type',
    markers=True,
    dashes=False
)
plt.title('Effect of Unemployment Rate on Vehicle Type Sales During Recessions')
plt.xlabel('Unemployment Rate')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()
```
