# Final Assignment: Part 1 - Create Visualizations using Matplotlib, Seaborn & Folium

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium

# Load data
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"
df = pd.read_csv(url)

# TASK 1.1: Annual automobile sales fluctuation
df_line = df.groupby('Year')['Automobile_Sales'].mean()
plt.figure(figsize=(10, 5))
df_line.plot(kind='line')
plt.title('Average Annual Automobile Sales Fluctuations')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.show()

# TASK 1.2: Sales by vehicle type during recession
df_rec = df[df['Recession'] == 1]
df_rec_type = df_rec.groupby(['Year', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_rec_type, x='Year', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Automobile Sales during Recession by Vehicle Type')
plt.show()

# TASK 1.3: Vehicle sales during recession vs non-recession
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Recession', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Vehicle-Wise Sales During Recession and Non-Recession Periods')
plt.xticks(ticks=[0, 1], labels=['Non-Recession', 'Recession'])
plt.show()

# TASK 1.4: Subplots for GDP during recession vs non-recession
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.lineplot(x='Year', y='GDP', data=df[df['Recession'] == 1])
plt.title('GDP Variation During Recession')
plt.subplot(1, 2, 2)
sns.lineplot(x='Year', y='GDP', data=df[df['Recession'] == 0])
plt.title('GDP Variation During Non-Recession')
plt.tight_layout()
plt.show()

# TASK 1.5: Seasonality impact on automobile sales
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Seasonality_Weight', hue='Seasonality_Weight')
plt.title('Seasonality Impact on Automobile Sales')
plt.show()

# TASK 1.6: Consumer confidence and automobile sales during recession
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df[df['Recession'] == 1], x='Consumer_Confidence', y='Automobile_Sales')
plt.title('Consumer Confidence vs Automobile Sales During Recession')
plt.show()

# TASK 1.7: Advertising expenditure correlation
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.scatterplot(data=df[df['Recession'] == 1], x='Advertising_Expenditure', y='Automobile_Sales')
plt.title('Recession: Ad Expenditure vs Sales')
plt.subplot(1, 2, 2)
sns.scatterplot(data=df[df['Recession'] == 0], x='Advertising_Expenditure', y='Automobile_Sales')
plt.title('Non-Recession: Ad Expenditure vs Sales')
plt.tight_layout()
plt.show()

# TASK 1.8: Pie chart for ad expenditure
exp_rec = df[df['Recession'] == 1].groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
exp_non_rec = df[df['Recession'] == 0].groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.pie(exp_rec, labels=exp_rec.index, autopct='%1.1f%%')
plt.title('Ad Expenditure Share: Recession')
plt.subplot(1, 2, 2)
plt.pie(exp_non_rec, labels=exp_non_rec.index, autopct='%1.1f%%')
plt.title('Ad Expenditure Share: Non-Recession')
plt.tight_layout()
plt.show()

# TASK 1.9: Unemployment rate effect
plt.figure(figsize=(10, 5))
sns.barplot(data=df[df['Recession'] == 1], x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Effect of Unemployment Rate on Vehicle Type and Sales')
plt.show()

```
